```python
import json
import pandas as pd
df=pd.read_json("HealthCareMagic-100k.json")

inputs = df["input"].tolist()
outputs = df["output"].tolist()
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
input_embeddings = model.encode(
    inputs,
    convert_to_tensor=True,
    show_progress_bar=True
)
```

```python
import torch
import pickle

# Save embeddings
torch.save(input_embeddings, "medical_embeddings.pt")

# Save outputs (doctor replies)
with open("medical_outputs.pkl", "wb") as f:
    pickle.dump(outputs, f)

# Save transformer model
model.save("medical_model")

print("Medical chatbot model saved")
```

```python
import torch
import pickle
from sentence_transformers import SentenceTransformer

model = SentenceTransformer()
input_embeddings = torch.load("medical_embeddings.pt")

with open("medical_outputs.pkl", "rb") as f:
    outputs = pickle.load(f)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def medical_chatbot(user_input):
    query_embedding = model.encode([user_input])
    similarities = cosine_similarity(query_embedding, input_embeddings)

    best_idx = np.argmax(similarities)
    best_score = similarities[0][best_idx]

    if best_score < 0.6:
        return "I recommend consulting a doctor for a proper medical evaluation."

    return outputs[best_idx]
```

```python
query = """
I feel dizzy when I stand up and the room spins.
I also feel nauseous and unsteady while walking.
"""

print(medical_chatbot(query))
```
